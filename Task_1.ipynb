```python
# Task 1: Data Visualization
# Prodigy InfoTech Data Science Internship

import pandas as pd
import matplotlib.pyplot as plt
import os

# Load the dataset
file_name = "API_SP.POP.TOTL_DS2_en_csv_v2_38144.csv"

if os.path.exists(file_name):
    df = pd.read_csv(file_name, skiprows=4) # Skip metadata header rows if necessary
    print(f"Successfully loaded local file: {file_name}")
elif 'df' in globals():
    print("Local CSV file not found. Reusing the 'df' DataFrame already loaded in memory.")
else:
    raise FileNotFoundError(f"Could not find '{file_name}' and no existing 'df' was found in memory. Please upload your CSV file.")

# Display the first five rows
print("\nFirst five rows of the dataset:")
print(df.head())

# Display column names
print("\nColumn names:")
print(df.columns.tolist())

# Choose a year to plot the population distribution (e.g., '2023' or '2024')
# Filter for actual countries (excluding regional aggregates if possible, or just plot the distribution)
column = "2023"

if column in df.columns:
    # Clean and convert chosen column to numeric, dropping missing values
    data = pd.to_numeric(df[column], errors="coerce").dropna()

    # Plotting the population distribution histogram
    plt.figure(figsize=(10, 6))
    plt.hist(data, bins=20, edgecolor="black", color="skyblue")
    plt.title(f"Distribution of Country Populations in {column}")
    plt.xlabel("Population")
    plt.ylabel("Number of Countries")
    plt.grid(axis="y", alpha=0.3)
    plt.ticklabel_format(style='plain', axis='x') # Avoid scientific notation on the X-axis
    plt.tight_layout()
    plt.show()
else:
    print(f"\n'{column}' column was not found in the dataset. Available numeric columns:")
    print(df.select_dtypes(include="number").columns.tolist())
```

```python

# Bar chart for categorical data

import matplotlib.pyplot as plt

column = "Gender"  # Replace with the actual column name

if column in df.columns:
    counts = df[column].value_counts()

    counts.plot(kind="bar", edgecolor="black", figsize=(7, 5))
    plt.title("Distribution of Gender")
    plt.xlabel("Gender")
    plt.ylabel("Count")
    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.show()
else:
    print("Available columns:", df.columns.tolist())
```
